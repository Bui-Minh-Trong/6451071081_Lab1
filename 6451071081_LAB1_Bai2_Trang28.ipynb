### Câu 1: Nhập dữ liệu và thống kê những thông tin cơ bản của các thuộc tính


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

# Khởi tạo DataFrame từ bảng mẫu 16 dòng được cung cấp trong tài liệu (trang 28-29)
data = {
    'Job': ['housemaid', 'services', 'blue-collar', 'admin.', 'blue-collar', 'services',
            'management', 'unemployed', 'technician', 'services', 'blue-collar',
            'self-employed', 'entrepreneur', 'services', 'technician', 'Retired'],
    'Marial': ['Married', 'Married', 'Divorced', 'Single', 'Single', 'Married',
               'Married', np.nan, 'Married', 'Divorced', 'Married',
               'Married', 'Married', 'Single', 'Married', 'Married'],
    'Education': ['basic.4y', 'high.school', 'basic.4y', 'high.school', 'basic.9y', 'high.school',
                  'high.school', 'university.degree', 'professional.course', 'high.school', 'high.school',
                  'basic.9y', 'professional.course', 'high.school', 'basic.9y', 'high.school'],
    'Default': ['no', 'unknown', 'unknown', 'no', 'no', 'no',
                'unknown', 'unknown', 'no', 'unknown', 'no',
                'no', 'no', 'no', 'no', 'no'],
    'Loan': ['no', 'no', 'no', 'no', 'no', 'no',
             'no', 'no', 'no', 'no', 'no',
             'no', 'yes', 'no', 'no', 'no'],
    'Duration': [261, 149, 1575, 338, 179, 1030,
                 149, 424, 1623, 568, 1297,
                 376, 576, 1059, 705, 532],
    'Cons.price.idx': [93.994, 93.994, 93.994, 93.994, 93.994, 93.994,
                       np.nan, 93.994, 93.994, 93.994, 93.994,
                       93.918, 93.994, 93.2, 93.2, 93.2],
    'Subscribe': ['no', 'no', 'yes', 'no', 'no', 'yes',
                  'no', 'no', 'yes', 'no', 'yes',
                  'no', 'no', 'yes', 'yes', 'no']
}
df_bank = pd.DataFrame(data)

print("Mẫu dữ liệu Bank Marketing (16 mẫu):")
display(df_bank)

print("\nCấu trúc dữ liệu (info):")
print(df_bank.info())

print("\nThống kê mô tả định lượng (describe):")
display(df_bank.describe())

```

**Nhận xét Câu 1:**

- Thuộc tính quyết định: `Subscribe` (khách hàng có đăng ký gửi tiền kỳ hạn hay không).
- Thuộc tính số: `Duration` (thời lượng liên hệ cuối cùng bằng giây), `Cons.price.idx` (chỉ số giá tiêu dùng).
- Thuộc tính định loại: `Job`, `Marial`, `Education`, `Default`, `Loan`.


### Câu 2: Thống kê số lượng giá trị rỗng (NaN) và giá trị chưa xác định ('unknown')


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

missing_summary = pd.DataFrame({
    'Số lượng NaN': df_bank.isnull().sum(),
    'Số lượng "unknown"': (df_bank == 'unknown').sum(),
    'Tổng giá trị thiếu': df_bank.isnull().sum() + (df_bank == 'unknown').sum()
})
print("Bảng thống kê giá trị rỗng và unknown:")
display(missing_summary)
```

**Nhận xét Câu 2:**
- `Marial`: Có 1 giá trị rỗng NaN ở hàng số 8.
- `Cons.price.idx`: Có 1 giá trị rỗng NaN ở hàng số 7.
- `Default`: Có 5 giá trị `'unknown'` (chiếm 31.25% tập mẫu). Cần được xử lý trong bước tiền xử lý.


### Câu 3: Khảo sát mối liên hệ giữa Nghề nghiệp (Job) và Quyết định gửi tiền (Subscribe)

- Các nghề như `housemaid`, `admin.`, `management`, `unemployed` đều trả lời `no` trong mẫu này.


### Câu 4: Khảo sát mối liên hệ giữa Thời lượng cuộc gọi (Duration) và Quyết định gửi tiền (Subscribe)


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

plt.figure(figsize=(8, 4))
sns.boxplot(x='Subscribe', y='Duration', data=df_bank, palette='pastel')
plt.title("Tương quan giữa Thời lượng cuộc gọi (Duration) và Subscribe")
plt.xlabel("Đăng ký gửi tiền (Subscribe)")
plt.ylabel("Thời lượng liên hệ (giây)")
plt.show()
```

**Nhận xét Câu 4:**

- Những khách hàng từ chối (`Subscribe = no`) có thời lượng cuộc gọi ngắn hơn rất nhiều (hầu hết < 600 giây).


### Câu 5: Khảo sát mối quan hệ giữa Tình trạng vay nợ (Loan, Default) và Subscribe


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(data=df_bank, x='Loan', hue='Subscribe', ax=axes[0], palette='Set1')
axes[0].set_title("Vay cá nhân (Loan) vs Subscribe")
axes[0].set_xlabel("Đã có khoản vay (Loan)")
axes[0].set_ylabel("Số lượng")

sns.scatterplot(data=df_bank, x='Duration', y='Cons.price.idx', hue='Subscribe', s=120, ax=axes[1], palette='coolwarm')
axes[1].set_title("Duration vs Cons.price.idx theo Subscribe")
axes[1].set_xlabel("Duration (giây)")
axes[1].set_ylabel("Cons.price.idx")
plt.tight_layout()
plt.show()
```

**Nhận xét Câu 5:**
- Khách hàng không có khoản vay nợ (`Loan = no`) có tỷ lệ đồng ý gửi tiền cao hơn hẳn. Khách hàng đã có khoản vay (`Loan = yes`) đều từ chối gửi tiền.
- Khi chỉ số giá tiêu dùng thấp hơn (`Cons.price.idx = 93.2`), tỷ lệ đồng ý gửi tiền tăng lên rõ rệt.


### Câu 6: Tiền xử lý dữ liệu Bank Marketing


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

df_clean_bank = df_bank.copy()

# 1. Điền giá trị rỗng cho Cons.price.idx bằng giá trị trung bình (mean)
df_clean_bank['Cons.price.idx'] = df_clean_bank['Cons.price.idx'].fillna(round(df_clean_bank['Cons.price.idx'].mean(), 3))

# 2. Điền giá trị rỗng cho Marial bằng mode
df_clean_bank['Marial'] = df_clean_bank['Marial'].fillna(df_clean_bank['Marial'].mode()[0])

# 3. Xử lý giá trị 'unknown' của cột Default
df_clean_bank['Default'] = df_clean_bank['Default'].replace('unknown', 'no')

# 4. Mã hóa nhãn nhị phân: Subscribe (yes: 1, no: 0)
df_clean_bank['Subscribe'] = df_clean_bank['Subscribe'].map({'yes': 1, 'no': 0})

# 5. Phân nhóm (Binning) thuộc tính Duration thành 3 nhóm: Short, Medium, Long
df_clean_bank['Duration_Group'] = pd.cut(df_clean_bank['Duration'], bins=[0, 300, 1000, np.inf], labels=['Short', 'Medium', 'Long'])

print("Dữ liệu Bank Marketing sau khi tiền xử lý:")
display(df_clean_bank)

print("\nKiểm tra lại số lượng giá trị rỗng:")
print(df_clean_bank.isnull().sum())

```

**Nhận xét Câu 6:**
- Dữ liệu đã được làm sạch hoàn toàn: Điền khuyết cho các cột số và cột phân loại, xử lý các giá trị không xác định 'unknown'.
- Đã chuẩn hóa nhãn quyết định thành 0 và 1, đồng thời chia nhóm thuộc tính thời lượng cuộc gọi phục vụ cho việc huấn luyện mô hình phân loại.
