### Câu 2: Liệt kê danh sách 5 thuộc tính có số lượng giá trị rỗng nhiều nhất


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

# Load dữ liệu Titanic
data_candidates = ['titanic.csv', '../titanic.csv', '../../TH1/titanic.csv', '../TH1/titanic.csv']
data_path = next((p for p in data_candidates if os.path.exists(p)), 'titanic.csv')
df = pd.read_csv(data_path)

missing_df = pd.DataFrame({
    'Số lượng rỗng': df.isnull().sum(),
    'Tỷ lệ rỗng (%)': (df.isnull().sum() / len(df)) * 100
})
top5_missing = missing_df.sort_values(by='Số lượng rỗng', ascending=False).head(5)
print("Top 5 thuộc tính có số lượng giá trị rỗng nhiều nhất:")
display(top5_missing)
```

3. boat: 823 giá trị rỗng (62.87%) - thuyền cứu sinh chỉ ghi nhận cho người sống sót.
