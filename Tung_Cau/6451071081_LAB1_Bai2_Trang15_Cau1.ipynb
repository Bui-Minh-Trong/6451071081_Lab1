### Câu 1: Câu 1: Nhập dữ liệu và thống kê những thông tin cơ bản của các thuộc tính


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

# 1. Đọc dữ liệu titanic.csv
import os
data_candidates = ['titanic.csv', '../TH1/titanic.csv', '../titanic.csv', 'datasets/titanic.csv']
data_path = next((p for p in data_candidates if os.path.exists(p)), 'titanic.csv')
df = pd.read_csv(data_path)

# Xem 5 dòng đầu
print("5 dòng đầu tiên:")
display(df.head())

# Xem thông tin cấu trúc
print("
Thông tin dữ liệu (info):")
print(df.info())

# Thống kê mô tả các thuộc tính định lượng
print("
Thống kê mô tả thuộc tính định lượng (describe):")
display(df.describe())

# Thống kê mô tả các thuộc tính định tính
print("
Thống kê mô tả thuộc tính định tính (describe category):")
display(df.describe(include=['O']))
```
