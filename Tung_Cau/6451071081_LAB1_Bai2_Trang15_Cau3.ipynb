### Câu 3: Vẽ biểu đồ khảo sát sự liên quan Độ tuổi (age), Giới tính (sex) với sống sót (survived)


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

# Load dữ liệu Titanic
data_candidates = ['titanic.csv', '../titanic.csv', '../../TH1/titanic.csv', '../TH1/titanic.csv']
data_path = next((p for p in data_candidates if os.path.exists(p)), 'titanic.csv')
df = pd.read_csv(data_path)

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))
women = df[df['sex'] == 'female']
men = df[df['sex'] == 'male']

sns.histplot(women[women['survived'] == 1].age.dropna(), bins=18, label='Sống sót', ax=axes[0], kde=False, color='green')
sns.histplot(women[women['survived'] == 0].age.dropna(), bins=40, label='Tử vong', ax=axes[0], kde=False, color='red')
axes[0].legend(); axes[0].set_title('Nữ giới (Female) - Sống sót theo tuổi'); axes[0].set_xlabel('Tuổi')

sns.histplot(men[men['survived'] == 1].age.dropna(), bins=18, label='Sống sót', ax=axes[1], kde=False, color='green')
sns.histplot(men[men['survived'] == 0].age.dropna(), bins=40, label='Tử vong', ax=axes[1], kde=False, color='red')
axes[1].legend(); axes[1].set_title('Nam giới (Male) - Sống sót theo tuổi'); axes[1].set_xlabel('Tuổi')
plt.tight_layout(); plt.show()
```

**Nhận xét Câu 3:**
- Nữ giới có cơ hội sống sót cao vượt trội ở hầu như mọi lứa tuổi (đặc biệt từ 14 đến 40 tuổi).
- Nam giới có tỷ lệ tử vong rất cao ở độ tuổi thanh niên (18-35 tuổi). Trẻ em trai từ 1 đến 10 tuổi có tỷ lệ sống sót cao hơn hẳn các nhóm tuổi nam khác.
