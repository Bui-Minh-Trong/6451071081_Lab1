### Câu 5: Đồ thị phân tán độ ẩm theo minutes


```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

cols = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']
data_candidates = ['tb_tracking.xlsx', '../tb_tracking.xlsx', '../../TH1/tb_tracking.xlsx', '../TH1/tb_tracking.xlsx']
data_path = next((p for p in data_candidates if os.path.exists(p)), 'tb_tracking.xlsx')
df_tracking = pd.read_excel(data_path, header=None, names=cols)
df_tracking['time'] = pd.to_datetime(df_tracking['time'])
df_clean_tracking = df_tracking[df_tracking['time'] >= '2018-01-03'].copy()
df_clean_tracking['minutes'] = (df_clean_tracking['time'].dt.hour * 60 + 
                                df_clean_tracking['time'].dt.minute + 
                                df_clean_tracking['time'].dt.second / 60)

plt.figure(figsize=(10, 5))
plt.scatter(df_clean_tracking['minutes'], df_clean_tracking['humidity'], alpha=0.3, c='royalblue', s=12)
plt.title("Đồ thị phân tán: Độ ẩm theo phút trong ngày")
plt.xlabel("Thời điểm trong ngày (phút)"); plt.ylabel("Độ ẩm (%)")
plt.grid(True, linestyle='--', alpha=0.6); plt.show()
```
