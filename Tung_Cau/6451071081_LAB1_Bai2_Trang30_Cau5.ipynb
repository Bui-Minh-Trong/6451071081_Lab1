### Câu 5: Câu 5 (Yêu cầu d - 2): Vẽ đồ thị phân tán (Scatter Plot) biểu diễn độ ẩm theo cột minutes


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

plt.figure(figsize=(10, 5))
plt.scatter(df_clean_tracking['minutes'], df_clean_tracking['humidity'], alpha=0.3, c='royalblue', s=12)
plt.title("Đồ thị phân tán: Độ ẩm theo thời điểm trong ngày (minutes)")
plt.xlabel("Thời điểm trong ngày (phút) - minutes")
plt.ylabel("Độ ẩm (humidity - %)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

**Nhận xét Câu 5:**
- Độ ẩm cao vào buổi sáng sớm và đêm muộn (phút 0 - 360 và 1200 - 1440). Vào khoảng giữa trưa và đầu giờ chiều (phút 600 - 900), độ ẩm giảm xuống thấp nhất tương ứng với sự gia tăng của nhiệt độ.
