### Câu 1: Đọc tập dữ liệu Groceries và mã hóa bằng TransactionEncoder


```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt
import os
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

data_candidates = ['groceries.csv', '../groceries.csv', '../../TH1/groceries.csv', '../TH1/groceries.csv']
data_path = next((p for p in data_candidates if os.path.exists(p)), 'groceries.csv')

dataset = []
with open(data_path, 'r', encoding='utf-8') as f:
    for line in f:
        transaction = [item.strip() for item in line.strip().split(',') if item.strip()]
        if transaction:
            dataset.append(transaction)

te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df_groceries = pd.DataFrame(te_ary, columns=te.columns_)
min_sups = [0.05, 0.04, 0.03, 0.02, 0.01]

print(f"Tổng số giao dịch: {len(dataset):,}")
print("Mẫu 3 giao dịch đầu:")
for i in range(3):
    print(f"Giao dịch {i+1}: {dataset[i]}")
print(f"\nKích thước ma trận One-Hot: {df_groceries.shape[0]:,} dòng x {df_groceries.shape[1]} cột")
display(df_groceries.iloc[:5, :8])
```
