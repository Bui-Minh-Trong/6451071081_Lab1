### Câu 4: Lập bảng so sánh thời gian thực thi Apriori vs FP-Growth


```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt
import os
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

data_candidates = ['groceries.csv', '../groceries.csv', '../../TH1/groceries.csv', '../TH1/groceries.csv']
data_path = next((p for p in data_candidates if os.path.exists(p)), 'groceries.csv')

dataset = []
with open(data_path, 'r', encoding='utf-8') as f:
    for line in f:
        transaction = [item.strip() for item in line.strip().split(',') if item.strip()]
        if transaction:
            dataset.append(transaction)

te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df_groceries = pd.DataFrame(te_ary, columns=te.columns_)
min_sups = [0.05, 0.04, 0.03, 0.02, 0.01]

time_apriori = []
time_fpgrowth = []
num_fi = []
for sup in min_sups:
    s1 = time.time(); fi1 = apriori(df_groceries, min_support=sup, use_colnames=True); t1 = time.time() - s1
    s2 = time.time(); fi2 = fpgrowth(df_groceries, min_support=sup, use_colnames=True); t2 = time.time() - s2
    time_apriori.append(t1); time_fpgrowth.append(t2); num_fi.append(len(fi1))

df_compare = pd.DataFrame({
    'min_sup (%)': [f"{s*100:.1f}%" for s in min_sups],
    'Số tập phổ biến': num_fi,
    'Thời gian Apriori (s)': [round(t, 4) for t in time_apriori],
    'Thời gian FP-Growth (s)': [round(t, 4) for t in time_fpgrowth],
    'Tốc độ chênh lệch (lần)': [round(ta / tf, 2) if tf > 0 else np.nan for ta, tf in zip(time_apriori, time_fpgrowth)]
})
display(df_compare)
```

**Nhận xét Câu 4:**
- FP-Growth luôn nhanh hơn Apriori ở mọi ngưỡng hỗ trợ.
