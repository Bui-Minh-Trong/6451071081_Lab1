### Câu 5: Vẽ biểu đồ so sánh thời gian xử lý


```python
import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt
import os
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

data_candidates = ['groceries.csv', '../groceries.csv', '../../TH1/groceries.csv', '../TH1/groceries.csv']
data_path = next((p for p in data_candidates if os.path.exists(p)), 'groceries.csv')

dataset = []
with open(data_path, 'r', encoding='utf-8') as f:
    for line in f:
        transaction = [item.strip() for item in line.strip().split(',') if item.strip()]
        if transaction:
            dataset.append(transaction)

te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df_groceries = pd.DataFrame(te_ary, columns=te.columns_)
min_sups = [0.05, 0.04, 0.03, 0.02, 0.01]

time_apriori = []
time_fpgrowth = []
for sup in min_sups:
    s1 = time.time(); fi1 = apriori(df_groceries, min_support=sup, use_colnames=True); t1 = time.time() - s1
    s2 = time.time(); fi2 = fpgrowth(df_groceries, min_support=sup, use_colnames=True); t2 = time.time() - s2
    time_apriori.append(t1); time_fpgrowth.append(t2)

plt.figure(figsize=(9, 5))
sup_percent = [s * 100 for s in min_sups]
plt.plot(sup_percent, time_apriori, marker='o', color='red', linewidth=2, label='Apriori')
plt.plot(sup_percent, time_fpgrowth, marker='s', color='green', linewidth=2, label='FP-Growth')
plt.title("So sánh thời gian xử lý giữa Apriori và FP-Growth")
plt.xlabel("min_sup (%)"); plt.ylabel("Thời gian (giây)")
plt.gca().invert_xaxis()
plt.legend(); plt.grid(True, linestyle='--', alpha=0.6); plt.show()
```

**Nhận xét Câu 5:**
- Đường của Apriori dốc đứng khi min_sup giảm; đường FP-Growth duy trì độ dốc rất êm.
