### Câu 4: Khảo sát Cổng lên tàu (embarked) và Loại vé (pclass) với sống sót theo Giới tính


```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

# Load dữ liệu Titanic
data_candidates = ['titanic.csv', '../titanic.csv', '../../TH1/titanic.csv', '../TH1/titanic.csv']
data_path = next((p for p in data_candidates if os.path.exists(p)), 'titanic.csv')
df = pd.read_csv(data_path)

FacetGrid = sns.FacetGrid(df, row='embarked', height=3, aspect=1.8)
FacetGrid.map(sns.pointplot, 'pclass', 'survived', 'sex', palette='deep', order=[1, 2, 3], hue_order=['female', 'male'])
FacetGrid.add_legend(); plt.show()
```

**Nhận xét Câu 4:**
- Nữ giới lên tàu ở cả 3 cổng đều có tỷ lệ sống sót vượt trội so với nam giới.
- Cổng C (Cherbourg): Nam giới mua vé hạng 1 có tỷ lệ sống sót cao hơn rõ rệt so với cổng S và Q.
- Loại vé: Ở cả 3 cổng, vé hạng 1 có tỷ lệ sống cao nhất; vé hạng 3 có tỷ lệ sống thấp nhất.
