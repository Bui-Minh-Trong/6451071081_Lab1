### Câu 1 (Yêu cầu a): Đọc dữ liệu và thêm tiêu đề các cột theo thứ tự: id, device_id, co_level, humidity, temperature, time


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

cols = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']
import os
data_candidates = ['tb_tracking.xlsx', '../TH1/tb_tracking.xlsx', '../tb_tracking.xlsx', 'datasets/tb_tracking.xlsx']
data_path = next((p for p in data_candidates if os.path.exists(p)), 'tb_tracking.xlsx')
df_tracking = pd.read_excel(data_path, header=None, names=cols)

print(f"Kích thước tập dữ liệu: {df_tracking.shape[0]:,} dòng x {df_tracking.shape[1]} cột")
print("5 dòng đầu tiên với tiêu đề cột đã đặt:")
display(df_tracking.head())

print("\nCấu trúc dữ liệu (info):")
print(df_tracking.info())

```

**Nhận xét Câu 1:**
- Đã gắn thành công tiêu đề cho 6 cột: `id`, `device_id`, `co_level`, `humidity`, `temperature`, `time`.
- Tổng cộng có 36,464 dòng dữ liệu cảm biến đo đạc môi trường nhà xưởng.


### Câu 2 (Yêu cầu b): Lọc bỏ dữ liệu sai lệch, chỉ giữ lại dữ liệu đo đạc bắt đầu từ ngày 03/01/2018


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

df_tracking['time'] = pd.to_datetime(df_tracking['time'])
df_clean_tracking = df_tracking[df_tracking['time'] >= '2018-01-03'].copy()

print(f"Số dòng dữ liệu ban đầu: {len(df_tracking):,}")
print(f"Số dòng sau khi lọc (>= 2018-01-03): {len(df_clean_tracking):,}")
print(f"Số dòng bị loại bỏ (trước 03/01/2018): {len(df_tracking) - len(df_clean_tracking):,}")

print("\n5 dòng đầu tiên sau khi lọc:")
display(df_clean_tracking.head())

```

**Nhận xét Câu 2:**
- Do cảm biến được hiệu chỉnh lại và bắt đầu đo chính xác từ ngày 03/01/2018, các bản ghi trước mốc thời gian này đã được loại bỏ hoàn toàn để tránh làm sai lệch mô hình.


### Câu 3 (Yêu cầu c): Bóc tách cột time thành thời điểm đo đạc trong ngày tính theo phút (minutes)


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

# Tính thời điểm trong ngày theo phút: giờ * 60 + phút + giây / 60
df_clean_tracking['minutes'] = (df_clean_tracking['time'].dt.hour * 60 + 
                                df_clean_tracking['time'].dt.minute + 
                                df_clean_tracking['time'].dt.second / 60)

print("Bảng dữ liệu sau khi bổ sung cột minutes:")
display(df_clean_tracking[['time', 'minutes', 'co_level', 'humidity', 'temperature']].head())

print("\nThống kê mô tả thuộc tính minutes:")
display(df_clean_tracking['minutes'].describe())

```

**Nhận xét Câu 3:**
- Cột `minutes` thể hiện thời điểm trong ngày (từ 0 đến 1440 phút). Biến đổi này giúp chuẩn hóa dữ liệu thời gian về cùng một chu kỳ 24 giờ trong ngày để dễ dàng quan sát sự biến thiên của các chỉ số môi trường.


### Câu 4 (Yêu cầu d - 1): Vẽ đồ thị phân tán (Scatter Plot) biểu diễn nồng độ khí CO theo cột minutes


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

plt.figure(figsize=(10, 5))
plt.scatter(df_clean_tracking['minutes'], df_clean_tracking['co_level'], alpha=0.3, c='crimson', s=12)
plt.title("Đồ thị phân tán: Nồng độ khí CO theo thời điểm trong ngày (minutes)")
plt.xlabel("Thời điểm trong ngày (phút) - minutes")
plt.ylabel("Nồng độ khí CO (co_level)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

**Nhận xét Câu 4:**
- Nồng độ CO tăng mạnh vào các khung giờ làm việc ban ngày (khoảng từ phút 480 tức 8h sáng đến phút 1020 tức 17h chiều) khi các máy móc và công nhân hoạt động tích cực trong nhà xưởng. Vào ban đêm, nồng độ CO giảm xuống mức thấp.


### Câu 5 (Yêu cầu d - 2): Vẽ đồ thị phân tán (Scatter Plot) biểu diễn độ ẩm theo cột minutes


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

plt.figure(figsize=(10, 5))
plt.scatter(df_clean_tracking['minutes'], df_clean_tracking['humidity'], alpha=0.3, c='royalblue', s=12)
plt.title("Đồ thị phân tán: Độ ẩm theo thời điểm trong ngày (minutes)")
plt.xlabel("Thời điểm trong ngày (phút) - minutes")
plt.ylabel("Độ ẩm (humidity - %)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

**Nhận xét Câu 5:**
- Độ ẩm cao vào buổi sáng sớm và đêm muộn (phút 0 - 360 và 1200 - 1440). Vào khoảng giữa trưa và đầu giờ chiều (phút 600 - 900), độ ẩm giảm xuống thấp nhất tương ứng với sự gia tăng của nhiệt độ.


### Câu 6 (Yêu cầu d - 3): Vẽ đồ thị phân tán (Scatter Plot) biểu diễn nhiệt độ theo cột minutes


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

plt.figure(figsize=(10, 5))
plt.scatter(df_clean_tracking['minutes'], df_clean_tracking['temperature'], alpha=0.3, c='darkorange', s=12)
plt.title("Đồ thị phân tán: Nhiệt độ theo thời điểm trong ngày (minutes)")
plt.xlabel("Thời điểm trong ngày (phút) - minutes")
plt.ylabel("Nhiệt độ (temperature - °C)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

**Nhận xét Câu 6:**
- Nhiệt độ có sự tương quan nghịch rõ rệt với độ ẩm: Nhiệt độ thấp nhất vào khoảng 5h - 6h sáng và đạt đỉnh cao nhất vào khoảng 12h - 14h trưa (phút 720 - 840).
