### Câu 1: Nhập dữ liệu và thống kê những thông tin cơ bản của các thuộc tính


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

# 1. Đọc dữ liệu titanic.csv
import os
data_candidates = ['titanic.csv', '../TH1/titanic.csv', '../titanic.csv', 'datasets/titanic.csv']
data_path = next((p for p in data_candidates if os.path.exists(p)), 'titanic.csv')
df = pd.read_csv(data_path)

# Xem 5 dòng đầu
print("5 dòng đầu tiên:")
display(df.head())

# Xem thông tin cấu trúc
print("\nThông tin dữ liệu (info):")
print(df.info())

# Thống kê mô tả các thuộc tính định lượng
print("\nThống kê mô tả thuộc tính định lượng (describe):")
display(df.describe())

# Thống kê mô tả các thuộc tính định tính
print("\nThống kê mô tả thuộc tính định tính (describe category):")
display(df.describe(include=['O']))

```

### Câu 3: Vẽ biểu đồ khảo sát sự liên quan giữa thuộc tính Độ tuổi (age), Giới tính (sex) và thuộc tính quyết định (survived) và đưa ra nhận xét


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))
women = df[df['sex'] == 'female']
men = df[df['sex'] == 'male']

# Biểu đồ cho Nữ giới
sns.histplot(women[women['survived'] == 1].age.dropna(), bins=18, label='Sống sót (Survived)', ax=axes[0], kde=False, color='green')
sns.histplot(women[women['survived'] == 0].age.dropna(), bins=40, label='Tử vong (Not survived)', ax=axes[0], kde=False, color='red')
axes[0].legend()
axes[0].set_title('Nữ giới (Female) - Tương quan tuổi & sống sót')
axes[0].set_xlabel('Tuổi (Age)')

# Biểu đồ cho Nam giới
sns.histplot(men[men['survived'] == 1].age.dropna(), bins=18, label='Sống sót (Survived)', ax=axes[1], kde=False, color='green')
sns.histplot(men[men['survived'] == 0].age.dropna(), bins=40, label='Tử vong (Not survived)', ax=axes[1], kde=False, color='red')
axes[1].legend()
axes[1].set_title('Nam giới (Male) - Tương quan tuổi & sống sót')
axes[1].set_xlabel('Tuổi (Age)')

plt.tight_layout()
plt.show()
```

**Nhận xét Câu 3:**
- Ở nữ giới: Cơ hội sống sót cao vượt trội ở hầu như mọi lứa tuổi, đặc biệt là nhóm từ 14 đến 40 tuổi.
- Ở nam giới: Tỷ lệ tử vong rất cao, nhất là nam thanh niên trong độ tuổi 18 đến 35 tuổi. Tuy nhiên, các bé trai từ 1 đến 10 tuổi có tỷ lệ sống sót cao hơn hẳn các nhóm tuổi nam khác, do chính sách ưu tiên trẻ em và phụ nữ.


### Câu 4: Vẽ biểu đồ khảo sát sự liên quan giữa thuộc tính Cổng lên tàu (embarked) và loại vé (pclass) với khả năng sống sót (survived), phân biệt theo Giới tính (sex) và đưa ra nhận xét


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

FacetGrid = sns.FacetGrid(df, row='embarked', height=3, aspect=1.8)
FacetGrid.map(sns.pointplot, 'pclass', 'survived', 'sex', palette='deep', order=[1, 2, 3], hue_order=['female', 'male'])
FacetGrid.add_legend()
plt.show()
```

**Nhận xét Câu 4:**
- Tại cả 3 cổng lên tàu (S, C, Q), tỷ lệ sống sót của nữ giới luôn cao hơn rất nhiều so với nam giới.
- Cổng C (Cherbourg): Tỷ lệ sống sót của nam giới mua vé hạng 1 cao hơn rõ rệt so với nam giới lên tàu ở cổng S và Q.
- Loại vé (pclass): Ở mọi cổng, hành khách mua vé hạng 1 có tỷ lệ sống sót cao nhất; hành khách mua vé hạng 3 có tỷ lệ sống sót thấp nhất.


### Câu 5: Thống kê số người đi cùng (anh chị em/vợ chồng hoặc cha mẹ/con cái) và khảo sát sự liên quan với thuộc tính quyết định và đưa ra nhận xét


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

# relatives = sibsp + parch
df['relatives'] = df['sibsp'] + df['parch']
df.loc[df['relatives'] > 0, 'not_alone'] = 1
df.loc[df['relatives'] == 0, 'not_alone'] = 0
df['not_alone'] = df['not_alone'].astype(int)

print("Số người đi một mình (not_alone = 0):", (df['not_alone'] == 0).sum())
print("Số người đi cùng người thân (not_alone = 1):", (df['not_alone'] == 1).sum())

plt.figure(figsize=(10, 4))
sns.pointplot(x='relatives', y='survived', data=df, color='royalblue')
plt.title("Tương quan giữa số lượng người thân đi cùng (relatives) và khả năng sống sót")
plt.xlabel("Số người thân đi cùng (relatives)")
plt.ylabel("Tỷ lệ sống sót (survived)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

**Nhận xét Câu 5:**
- Số người đi một mình là 790 người; số người đi cùng người thân là 519 người.
- Những hành khách có từ 1 đến 3 người thân đi cùng có tỷ lệ sống sót cao nhất (từ 50% đến hơn 70%).
- Những người đi một mình có tỷ lệ sống sót thấp hơn (~30%).
- Những gia đình quá đông người (từ 4 đến 6 người, đặc biệt 7-10 người) có tỷ lệ sống sót thấp nhất (gần 0%) do khó khăn trong việc tìm kiếm, tập hợp gia đình để lên thuyền cứu sinh.


### Câu 6: Tiền xử lý các thuộc tính thiếu dữ liệu và xóa bỏ các thuộc tính không cần thiết


```python
print("Họ và tên: Bùi Minh Trọng - MSSV: 6451071081")

# 1. Xóa các thuộc tính không cần thiết, rò rỉ dữ liệu hoặc thiếu quá nhiều
drop_cols = ['ticket', 'cabin', 'boat', 'body', 'home.dest']
df_clean = df.drop(columns=[col for col in drop_cols if col in df.columns]).copy()

# 2. Điền giá trị thiếu cho thuộc tính tuổi (age) bằng giá trị ngẫu nhiên trong khoảng [mean - std, mean + std]
np.random.seed(42)
mean_age = df_clean['age'].mean()
std_age = df_clean['age'].std()
is_null_age = df_clean['age'].isnull().sum()
rand_age = np.random.randint(int(mean_age - std_age), int(mean_age + std_age), size=is_null_age)
age_slice = df_clean['age'].copy()
age_slice[np.isnan(age_slice)] = rand_age
df_clean['age'] = age_slice.astype(int)

# 3. Điền giá trị thiếu cho cổng lên tàu (embarked) bằng giá trị mode
common_embarked = df_clean['embarked'].mode()[0]
df_clean['embarked'] = df_clean['embarked'].fillna(common_embarked)

# 4. Điền giá trị thiếu cho giá vé (fare) bằng median
df_clean['fare'] = df_clean['fare'].fillna(df_clean['fare'].median()).astype(int)

print("Kiểm tra lại số lượng giá trị thiếu sau khi tiền xử lý:")
print(df_clean.isnull().sum())
print("\n5 dòng dữ liệu sau tiền xử lý:")
display(df_clean.head())

```

**Nhận xét Câu 6:**
- Đã loại bỏ hoàn toàn các cột không cần thiết (`ticket`, `cabin`, `boat`, `body`, `home.dest`).
- Đã điền khuyết cho các cột `age`, `embarked`, `fare`.
- Tập dữ liệu hiện tại sạch hoàn toàn (0 giá trị NaN) và sẵn sàng cho các bước mô hình hóa khai phá dữ liệu.
